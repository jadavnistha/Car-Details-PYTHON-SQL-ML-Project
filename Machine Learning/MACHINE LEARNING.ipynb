```python
x=data.drop("price_category",axis=1)
```

```python
y=data["price_category"]
```

```python
from sklearn.preprocessing import LabelEncoder,StandardScaler
```

```python
encode=LabelEncoder()
```

```python
data["vehicle_uid"]=encode.fit_transform(data["vehicle_uid"])
```

```python
data
```

```python
x=x.apply(encode.fit_transform)
```

```python
scale=StandardScaler()
```

```python
scale.fit_transform(x)
```

```python
x=pd.DataFrame(scale.fit_transform(x),columns=x.columns)
```

```python
x
```

```python
import seaborn as sns
```

```python
sns.pairplot(x)
```

```python
from sklearn.model_selection import  train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=42,train_size=0.7)
```

```python
from sklearn.svm import SVC
```

```python
model=SVC(C=90)
```

```python
model.fit(x_train,y_train)
```

```python
pri=model.predict(x_test)
```

```python
model.score(x_train,y_train)*100
```

```python
model.score(x_test,y_test)*100
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
accuracy_score(y_test,pri)
```

```python
c=confusion_matrix(y_test,pri)
```

```python
sns.heatmap(c,annot=True)
```

```python
print(classification_report(y_test,pri))
```
